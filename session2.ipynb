```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.visualization import plot_bloch_multivector
from IPython.display import display # Import display

# Crear circuito con un solo qubit
qc = QuantumCircuit(1)
# Pero ese qubit debe empezar con |0> (no hace falta escribir explícitamente |0>, porque Qiskit inicializa los qubits en ∣0⟩)
# Damos al usuario para que elija cual opcion quiere
print("Choose a gate:")
print("1. X")
print("2. Y")
print("3. Z")
print("4. H")
print("5. RY")

# Opciones faciles
while True:
    choice = input("Enter your choice: ")

    if choice in ["1", "2", "3", "4", "5"]:
        break
    else:
        print("Invalid option. Please choose a number from 1 to 5.")
if choice == "1":
    qc.x(0)

elif choice == "2":
    qc.y(0)

elif choice == "3":
    qc.z(0)

elif choice == "4":
    qc.h(0)
# Opcion 5 primero pedirle el ángulo
elif choice == "5":
    theta = float(input("Enter the angle in radians: "))
    qc.ry(theta, 0)
# Dibujar circuito
fig = qc.draw("mpl")
display(fig) # Changed to display(fig)
plt.close(fig) # Close the figure to prevent it from showing up in other cells
```
